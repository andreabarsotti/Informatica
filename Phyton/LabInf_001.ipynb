Ci chiediamo se dato $x$ numero floating point, anche $1 / x$ lo è, e se viene calcolato esattamente dal processore. Possiamo verificarlo a mano per qualche caso di interesse, e poi verificare quali fra gli interi $1, 2, \ldots, 100$ hanno inversa in floating point.

````python
y = 1 / x
````

````python
x*y == 1.0
````

````python
for j in range(1, 101):
    x = np.float64(j)
    y = 1 / x
    if x*y != 1.0:
        print("%f non ha inversa esatta in floating point" % x)
        print(x*y - 1.0)
````

## Calcolo dell'esponenziale
Un buon esempio di quando si incontrano problemi di cancellazione numerica si ha con il calcolo dell'esponenziale. 
Come possiamo calcolare $e^x$ usando solo le operazioni aritmetiche? Un modo naturale è la serie di Taylor:
$$
  e^x \approx 1 + x + \frac{x^2}{2} + \ldots + \frac{x^k}{k!}
$$
La funzione ```myexp``` implementa questao idea, troncando la serie all'ordine $k$.

````python
def myexp(x, k):
    y = 0
    w = 1
    for j in range(k):
        y = y + w
        w = w * x / (j + 1)
    return y
````

Controlliamo l'errore relativo su qualche numero $x > 0$; ad esempio $x = 10$:

````python
( myexp(10, 50) - np.exp(10) ) / np.exp(10)
````

Le cose non vanno altrettanto bene se prendiamo $x < 0$; come mai? cosa sta andando storto?

````python
( myexp(-15, 200) - np.exp(-15) ) / np.exp(-15)
````

Ce ne possiamo rendere conto controllando l'operazione
$$
  r = x + y + z, 
$$
costruita in modo che $r \ll x + y$. Come abbiamo visto a lezione, questo ci dice che l'errore relativo sul risultato potrebbe essere molto grande!

````python
x = 1.0
y = 1e-10 * np.random.randn()
z = -1.0

r = x + y + z
````

````python
np.abs(r - y) / np.abs(y)
````

Il problema si verifica anche nel calcolo di $e^x$ con la serie di Taylor se $x < 0$: stiamo sommando vari numeri 
con segni discordi, per ottenere un risultato molto piccolo. Per ovviare al problema, possiamo ricorrere al trucco 
di rappresentare $e^x = (e^{-x})^{-1}$:

````python
def myexp2(x, k):
    if x > 0:
        return myexp(x, k)
    else:
        return 1.0 / myexp(-x, k)
````

````python
( myexp2(-15, 200) - np.exp(-15) ) / np.exp(-15)
````
